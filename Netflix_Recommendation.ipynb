## Text_Preprocessing

```python
A['tags'] = A['overview'].fillna(' ')+A['keywords'].fillna(' ')+A['genres'].fillna(' ')
```

```python
A['tags']
```

## Lower

```python
def a(n):
    n = n.lower()
    return n
```

```python
A['tags'] = A['tags'].apply(a)
```

```python
A['tags']
```

## Stemming

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
from nltk.stem import WordNetLemmatizer
lemm = WordNetLemmatizer()
lemm.lemmatize('changed')
```

```python
def stemming(n):
    y = []
    for i in n.split():
        y.append(ps.stem(n))
    return  ' '.join(y)
```

```python
A['tags']
```

```python
A['tags'] = A['tags'].apply(stemming)
```

```python
A['tags']
```

## Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
tf = TfidfVectorizer(stop_words = 'english',max_features = 21000)
```

```python
vector = tf.fit_transform(A['tags'])
```

```python
tf.get_feature_names_out().tolist()
```

## Caluculate the Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity1 = pd.DataFrame(similarity,index = A['title'],columns = A['title'])
similarity1
```

```python
import numpy as np

def recommend(movie):
    idx = A[A['title'] == movie].index[0]
    movie_list = np.argsort(similarity1.iloc[idx][::-1][1:6])
    return A.iloc[movie_list].title.tolist()
```

```python
recommend('Interstellar')
```
